```python
from nhlpy import NHLClient

client = NHLClient(debug=True)

teams = client.teams.teams()
print(type(teams), teams[0])            # find the team abbreviation key

sched = client.schedule.team_season_schedule(team_abbr="MTL", season="20252026")
print(sched.keys())
print(sched["games"][0])                # find the game id, gameState, gameType keys

gid = sched["games"][0]["id"]           # adjust if the key differs
for name in ("play_by_play", "boxscore"):
    resp = getattr(client.game_center, name)(game_id=gid)
    print(name, list(resp.keys()))
```
